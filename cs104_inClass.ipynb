```python
count_max = int(input("Input number of grades here: "))
count = 0
sum = 0
while 1:
  if count == count_max:
    break

  try:
    grade = float(input("Input grade here: "))
    sum += grade
  except ValueError:
    print("invalid input")
    continue

  if grade >= 90:
    letter_grade = "A"
  elif grade >= 80:
    letter_grade = "B"
  elif grade >= 70:
    letter_grade = "C"
  elif grade >= 60:
    letter_grade = "D"
  else:
    letter_grade = "F"
  print("Your grade is: " + letter_grade)
  count += 1

print(f"Average grade is: {(sum/count):.2f}")
```
