```python
import pygame

# Initialize Pygame
pygame.init()

# Screen dimensions
SCREEN_WIDTH = 800
SCREEN_HEIGHT = 600

# Create the screen
screen = pygame.display.set_mode((SCREEN_WIDTH, SCREEN_HEIGHT))

# Title and Icon
pygame.display.set_caption("Space Invaders")
# You can add an icon here if you have an image file
# icon = pygame.image.load('invader.png')
# pygame.display.set_icon(icon)

# Game Loop
running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False

    # Fill the background (RGB color)
    screen.fill((0, 0, 0)) # Black background

    # Update the display
    pygame.display.flip()

pygame.quit()
```
