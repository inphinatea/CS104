```python
import math
def deg2rad(angle):
  radian = angle * math.pi / 180
  return radian

def my_function(x, y):
  z = math.sin(x) + math.cos(y)
  return z

angle1 = float(input("Enter the first angle in degrees: "))
angle2 = float(input("Enter the second angle in degrees: "))

angle1_rad = deg2rad(angle1)
angle2_rad = deg2rad(angle2)
result = my_function(angle1_rad, angle2_rad)
print("The result is:", round(result,2))
```

```python

def FtoC(F):
  C = (F - 32) * (5/9)
  return C

TempC = float(input("Enter temperature in Farenheit here: "))
TempF = FtoC(TempC)

print("The temperature in Celsius is", round(TempF,2))
```
